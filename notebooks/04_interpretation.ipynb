```python
import sys
from pathlib import Path

project_root = Path.cwd().parent
sys.path.append(str(project_root))

from src.geospatial import table2raster
from src.plots import plot_cluster_heatmap, plot_cluster_boxplots, plot_cluster_distribution

from main import DIR_PARIKA, DIR_ILMATSALU, INDEX_NAMES, OUTPUT_DIR_PLOTS, REGION_NAME_ILMATSALU, REGION_NAME_PARIKA, SRC_IMG_DRONE_NAME, SRC_IMG_S2_NAME
```

```python
clustered_file_parika_drone = r"..\results\Parika\drone_img\clustering\clustered_df.parquet"
reference_tif_parika_drone = r"..\results\Parika\drone_img\indices\ndvi.tif"

clustered_file_ilmatsalu_drone = r"..\results\Ilmatsalu\drone_img\clustering\clustered_df.parquet"
reference_tif_ilmatsalu_drone = r"..\results\Ilmatsalu\drone_img\indices\ndvi.tif"

clustered_file_parika_s2 = r"..\results\Parika\S2\clustering\clustered_df.parquet"
reference_tif_parika_s2 = r"..\results\Parika\S2\indices\ndvi.tif"

clustered_file_ilmatsalu_s2 = r"..\results\Ilmatsalu\S2\clustering\clustered_df.parquet"
reference_tif_ilmatsalu_s2 = r"..\results\Ilmatsalu\S2\indices\ndvi.tif"
```

```python
datasets = [
    {
        "region_name": REGION_NAME_PARIKA,
        "output_dir": DIR_PARIKA,
        "src_img": SRC_IMG_DRONE_NAME,
        "clustered_file": clustered_file_parika_drone,
        "reference_tif": reference_tif_parika_drone
    },
    {
        "region_name": REGION_NAME_ILMATSALU,
        "output_dir": DIR_ILMATSALU,
        "src_img": SRC_IMG_DRONE_NAME,
        "clustered_file": clustered_file_ilmatsalu_drone,
        "reference_tif": reference_tif_ilmatsalu_drone
    },
    {
        "region_name": REGION_NAME_PARIKA,
        "output_dir": DIR_PARIKA,
        "src_img": SRC_IMG_S2_NAME,
        "clustered_file": clustered_file_parika_s2,
        "reference_tif": reference_tif_parika_s2
    },
    {
        "region_name": REGION_NAME_ILMATSALU,
        "output_dir": DIR_ILMATSALU,
        "src_img": SRC_IMG_S2_NAME,
        "clustered_file": clustered_file_ilmatsalu_s2,
        "reference_tif": reference_tif_ilmatsalu_s2
    }
]


for dataset in datasets:

    print(f'Processing {dataset["region_name"]} - {dataset["src_img"]}')

    cluster_raster = table2raster(
        clustered_file=dataset["clustered_file"],
        reference_tif=dataset["reference_tif"],
        output_dir=dataset["output_dir"],
        src_img=dataset["src_img"]
    )

    cluster_profiles = plot_cluster_heatmap(
        clustered_file=dataset["clustered_file"],
        output_dir=OUTPUT_DIR_PLOTS,
        region_name=dataset["region_name"],
        src_img=dataset["src_img"]
    )

    plot_cluster_boxplots(
        clustered_file=dataset["clustered_file"],
        output_dir=OUTPUT_DIR_PLOTS,
        indices=INDEX_NAMES,
        region_name=dataset["region_name"],
        src_img=dataset["src_img"]
    )

    plot_cluster_distribution(
        clustered_file=dataset["clustered_file"],
        output_dir=OUTPUT_DIR_PLOTS,
        region_name=dataset["region_name"],
        src_img=dataset["src_img"]
    )
```
